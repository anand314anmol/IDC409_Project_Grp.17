```python
# Import necessary libraries
import pandas as pd

# 1. Load the dataset (using sep=';' because the file uses semicolons)
# Make sure 'student-mat.csv' is saved in the exact same folder as this notebook!
df = pd.read_csv('student-mat.csv', sep=';')

# 2. Check for any missing values in the dataset
print("Total missing values:", df.isnull().sum().sum())

# 3. Separate Features (X) and Target (y)
# Target (y): G3 is the final grade we want to predict
# Features (X): All other columns used to make the prediction
X = df.drop(columns=['G3'])
y = df['G3']

# 4. Convert categorical text columns (like 'yes'/'no', 'F'/'M') into numbers (One-Hot Encoding)
X = pd.get_dummies(X, drop_first=True)

print("Data preparation complete!")
print(f"Features shape (X): {X.shape}")
print(f"Target shape (y): {y.shape}")
```
